### aggregate function = to display single values instead of multiple values
often used with group by

```python
import pandas as pd
df = pd.read_csv("data.csv")
print(df.mean(numeric_only = True))
print("total")
print(df.sum(numeric_only = True))
print("min and max")
print(df.min(numeric_only = True))
print(df.max(numeric_only = True))
print("no. of count:")
print(df.count())

# for selecting a single column
print(df["Height"].sum())#no need bcoz only contains 
print(df["Type2"].count())
print(df["Legendary"].count())

```

### group by
