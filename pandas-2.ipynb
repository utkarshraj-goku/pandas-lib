```python
import pandas as pd

data = {
"Name": ["Spongebob", "Patrick", "squidward"],
"Age": [30, 35, 50]
}
#custom index
df = pd.DataFrame(data, index=["employee 1", "employee 2", "employee 3"])
print(df.iloc[2])

#add a new column
df["job"] = ["cook", "N/A","cashier"]

#add a new row
new_row = pd.DataFrame([{"Name": "rishu", "Age": 28, "Job": "Engineer"}],index=["employee 4"])
df = pd.concat([df, new_row])

print(df)
```
