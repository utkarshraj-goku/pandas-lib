### Data Cleaning

# data cleaning = it is the process of removing unnecessary data  ~75% of work done is data cleaning

```python
import pandas as pd
df = pd.read_csv("data.csv")
```

```python
# 1. drop irrevelant colums
df = df.drop(columns=["Legendary", "No"])
print(df)
```

### handle missing data

```python
#df = df.dropna(subset=["Type2"])
#print(df.to_string())
#dropna = drop not availabe
df = df.fillna({"Type2": "lawra"})
#fill not availaable with some assigned vlues
print(df.to_string())
```

### incosistent values

```python
df["Type1"] = df["Type1"].replace({"Grass": "GRASS",
                                    "Fire": "FIRE",
                                    "Water": "WATER"           })

print(df.to_string())
```

### standardized text 

```python
df["Name"] = df["Name"].str.lower()
print(df.to_string())
```

### 5. Fix data types

```python
df["Legendary"] = df["Legendary"].astype(bool)

print(df.to_string())
```

### remove duplivate data

```python
df = df.drop_duplicates()

print(df.to_string())
```
