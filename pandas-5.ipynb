### filtering

```python
import pandas as pd
df = pd.read_csv("data.csv")

#tall_pokemon = df[df["Height"]>=2]
#legendary_pokemon = df[df["Legendary"]==1]
#heavy_pokemon = df[df["Weight"]>=100]
water_pokemon = df[(df["Type2"] == "Water") |
(df["Type1"] == "Water")]
print(water_pokemon)
```

### and_operator

```python
ff_pokemon = df[(df["Type1"] == "Fire")&
(df["Type2"] == "Flying")]

print(ff_pokemon)
```
