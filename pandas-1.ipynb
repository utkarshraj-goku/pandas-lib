### series(1-d array)

```python
import pandas as pd

data = [False, True, False] #dtype in output can be changed by changing data type in this series

series = pd.Series(data)

print(series)
```

### custom indexing

```python
import pandas as pd

data = [100, 102, 104]

series = pd.Series(data, index=["a","b","#appartment3"] )
print(series)
print("for printing a single index's value",series.loc["b"])
print("for printing a integer value",series.iloc[0])
```

### filter with vaues

```python
data = [100, 102, 104, 200, 202]

series = pd.Series(data, index =["a", "b", "c","d","e"])

print(series[series < 200])
```

```python
import pandas as pd

calories = {"Day1": 1750, "Day 2": 2100,"Day 3": 2200}
series = pd.Series(calories)
print(series.loc["Day 3"])
series.loc["Day 3"] += 500
print(series)
#subscript operator
print(series[series >= 2000])
```
